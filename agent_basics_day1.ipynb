```python
import json, re

def add(a, b):
    return a + b

def multiply(a, b):
    return a * b

def salary_date():
    return "हर महीने की 5 तारीख"

tools = {"add": add, "multiply": multiply, "salary_date": salary_date}

def decide(question):
    nums = [int(n) for n in re.findall(r"\d+", question)]
    if "जोड़" in question:
        if len(nums) < 2:
            return {"final": "जोड़ने के लिए दो नंबर चाहिए"}
        return {"tool": "add", "args": {"a": nums[0], "b": nums[1]}}
    if "गुणा" in question:
        if len(nums) < 2:
            return {"final": "गुणा के लिए दो नंबर चाहिए"}
        return {"tool": "multiply", "args": {"a": nums[0], "b": nums[1]}}
    if "वेतन" in question:
        return {"tool": "salary_date", "args": {}}
    return {"final": "मुझे नहीं पता"}
```

```python
def run_agent(question):
    # 1. फ़ैसला लें कि कौन सा टूल इस्तेमाल करना है या सीधे जवाब देना है
    decision = decide(question)

    # 2. अगर सीधे फाइनल जवाब मिल गया है
    if "final" in decision:
        return decision["final"]

    # 3. अगर किसी टूल को कॉल करना है
    if "tool" in decision:
        tool_name = decision["tool"]
        args = decision.get("args", {})

        if tool_name in tools:
            # टूल को कॉल करें
            result = tools[tool_name](**args)
            return f"टूल '{tool_name}' का परिणाम: {result}"
        else:
            return f"त्रुटि: टूल '{tool_name}' उपलब्ध नहीं है।"

    return "कुछ गड़बड़ हुई।"
```

```python
# अलग-अलग इनपुट के साथ एजेंट का परीक्षण (Testing)
print(run_agent("5 और 10 को जोड़ो"))
print(run_agent("4 और 3 का गुणा करो"))
print(run_agent("मेरा वेतन कब आएगा?"))
print(run_agent("मौसम कैसा है?"))
```
